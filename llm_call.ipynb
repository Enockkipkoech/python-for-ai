```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from google import genai

client = genai.Client()

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="Tell me who invented AI, the so called 'fathers of ai' "
)

print(response.text)
```

```python
def generate_recipe(ingredients, cuisine,diet):
    
    # '''
    # generate a recipe using incredients provided. Cuisine can be "Italian", "Mexican", "Indian", "Kenyan" etc
    # diet can be "vegan", 'Vegeterian", "Jain", "Any", etc.
    # '''

    prompt = f'''
    Generate a one food/meal recipe using the ingredients: {','.join(ingredients)}
    Recipe should be not more than 100 words
    Cuisine:{cuisine}
    Diet:{diet}
    '''
    response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents=prompt
    )
    return response.text    
```

```python
recipe = generate_recipe(["tomatoes","onions","rice","salt"], "Kenyan", "vegan")
recipe
```
