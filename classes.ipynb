```python
from datetime import date

class Human:
    def __init__(this,name, birth_year):
        this.name = name
        this.birth_year = birth_year
        
    def get_age(this):
        current_year = date.today().year
        return current_year - this.birth_year

    def introduce(this):
        print(f"Hi, I'm {this.name}. My age is {this.get_age()} years")


Enock = Human("Enock", 1999)
Enock.get_age()
        
```
