```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()

with open("notes.text", "r") as f:
    document = f.read()

prompt = f"""
Here is my transcript: {document}
summarize it in 3 short bullet points.

"""

try:
    response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=prompt
    )
    summary=response.text
    
except Exception as e:
    summary = f"something went wrong!:({e})"

print(summary)


```

```python
with open("summary.text", "w", encoding="utf8") as f:
    f.write(summary)

print("summary saved to summary.text  file")
```
