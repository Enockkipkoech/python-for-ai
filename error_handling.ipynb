```python
from dotenv import load_dotenv
load_dotenv()

from google import genai

client = genai.Client()


def generate_recipe(ingredients, cuisine,diet):
    
    # '''
    # generate a recipe using incredients provided. Cuisine can be "Italian", "Mexican", "Indian", "Kenyan" etc
    # diet can be "vegan", 'Vegeterian", "Jain", "Any", etc.
    # '''

    prompt = f'''
    Generate a one food/meal recipe using the ingredients: {','.join(ingredients)}
    Recipe should be not more than 100 words
    Cuisine:{cuisine}
    Diet:{diet}
    '''

    try:
        response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=prompt
        )
        if not response.text:
            return "The model returned an empty response. PLease try again!"
                
        return response.text 
    except Exception as e:
        return f"Sorry, could not generate recipe! Error: ({e})"

print(generate_recipe(["tomatoes","onions","rice","salt"], "Kenyan", "vegan"))

    

    
```
