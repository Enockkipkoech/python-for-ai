```python
import sqlite3

connection = sqlite3.connect("shop.db")

cursor = connection.cursor()

print("database connected successfully!")

```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS reviews (id INTEGER PRIMARY KEY, text TEXT,label TEXT, score INTEGER)
"""    
)
```

```python
cursor.execute("INSERT INTO reviews (text,label,score) VALUES (?,?,?)",
               ("The retaurant was amazing & cool", "positive", 4))

connection.commit()

print("Review saved!")
```

```python
connection.commit()
```

```python
more_reviews = [
    ("Delivery was very late and food was cold.", "negative", 1),
    ("Good taste, fair price.", "positive", 4),
    ("It was okay, nothing special.", "neutral", 3),
    ("App kept crashing during payment.", "negative", 2),
]


cursor.executemany("INSERT INTO reviews (text,label,score) VALUES (?,?,?)", 
more_reviews)

connection.commit()

print(f"Added: {len(more_reviews)} more reviews.")
```

```python
cursor.execute("SELECT * FROM reviews")
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute("SELECT text, score FROM reviews WHERE label = ?", ("negative",))

negatives = cursor.fetchall()

print("Negative Reviews:")

for text, score in negatives:
    print(text, "(score:",score)

connection.close()    

print("db closed")
```
